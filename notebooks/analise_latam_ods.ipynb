## 1. Importação das bibliotecas


```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt

pd.set_option("display.max_columns", None)
pd.set_option("display.float_format", lambda x: f"{x:,.2f}")
```

## 2. Carregamento da base

No GitHub, o CSV está armazenado na pasta `dados`.  
Ao abrir este notebook no Google Colab, faça o upload do CSV para a sessão ou ajuste o caminho conforme necessário.


```python
# Opção para uso dentro do repositório:
arquivo = "../dados/LATAM_TAM_2019_2025_FINAL.csv"

# No Google Colab, se você enviar o CSV diretamente para a sessão, use:
# arquivo = "/content/LATAM_TAM_2019_2025_FINAL.csv"

df = pd.read_csv(arquivo, sep=None, engine="python", encoding="utf-8-sig")
df.head()
```

## 3. Conferência inicial da base


```python
print("Quantidade de linhas:", df.shape[0])
print("Quantidade de colunas:", df.shape[1])

display(df.head())
display(df.tail())
```

## 4. Verificação da qualidade dos dados

Nesta etapa são observados valores ausentes e registros duplicados antes das análises.


```python
# Valores ausentes por coluna
nulos = df.isna().sum().sort_values(ascending=False)
nulos[nulos > 0]
```

```python
# Quantidade de linhas duplicadas
print("Linhas duplicadas:", df.duplicated().sum())
```

## 5. Limpeza básica

São removidos espaços extras dos nomes das colunas e registros completamente duplicados, preservando os dados originais relevantes.


```python
df.columns = df.columns.str.strip()
df = df.drop_duplicates().copy()

print("Dimensão após a limpeza:", df.shape)
```

## 6. Identificação das variáveis disponíveis

Esta etapa ajuda a reconhecer quais variáveis da ANAC podem ser utilizadas para construir indicadores associados aos ODS 9 e 13.


```python
for coluna in df.columns:
    print(coluna)
```

## 7. Preparação temporal

O código procura automaticamente uma coluna relacionada a ano e, quando encontrada, apresenta a quantidade de registros por período.


```python
colunas_ano = [c for c in df.columns if "ano" in c.lower()]
print("Possíveis colunas de ano:", colunas_ano)

if colunas_ano:
    coluna_ano = colunas_ano[0]
    display(df[coluna_ano].value_counts(dropna=False).sort_index())
```
