## Ciência de dados
</div>

Com o Colab, você pode aproveitar todo o potencial das conhecidas bibliotecas Python para analisar e ver dados. A célula de códigos abaixo usa <strong>numpy</strong> para gerar dados aleatórios e <strong>matplotlib</strong> para visualizá-los. Para editar o código, basta clicar na célula e começar a editar.

É possível importar para os notebooks do Colab os dados da sua conta do Google Drive, como planilhas. Também é possível importar do GitHub e de muitas outras fontes. Para saber mais sobre como importar dados e como o Colab pode ser usado para a ciência de dados, consulte o link abaixo em <a href="#working-with-data">Como trabalhar com dados</a>.

```
import numpy as np
import IPython.display as display
from matplotlib import pyplot as plt
import io
import base64

ys = 200 + np.random.randn(100)
x = [x for x in range(len(ys))]

fig = plt.figure(figsize=(4, 3), facecolor='w')
plt.plot(x, ys, '-')
plt.fill_between(x, ys, 195, where=(ys > 195), facecolor='g', alpha=0.6)
plt.title("Sample Visualization", fontsize=10)

data = io.BytesIO()
plt.savefig(data)
image = F"data:image/png;base64,{base64.b64encode(data.getvalue()).decode()}"
alt = "Sample Visualization"
display.display(display.Markdown(F"""![{alt}]({image})"""))
plt.close(fig)
```

```
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt

print("Bibliotecas importadas com sucesso!")
```

```
arquivo = "/content/LATAM_TAM_2019_2025_FINAL.csv"

df = pd.read_csv(
    arquivo,
    sep=";",
    encoding="utf-8-sig"
)

print("Dataset carregado com sucesso!")
print("Linhas:", df.shape[0])
print("Colunas:", df.shape[1])
```

```
df.head()
```

```
print("Colunas disponíveis no dataset:\n")

for coluna in df.columns:
    print(coluna)
```

```
print("Informações gerais do dataset:\n")
df.info()
```

```
valores_ausentes = df.isnull().sum()

print("Valores ausentes por coluna:\n")
print(valores_ausentes)
```

```
percentual_ausentes = (df.isnull().sum() / len(df) * 100).round(2)

tabela_ausentes = pd.DataFrame({
    "Valores Ausentes": df.isnull().sum(),
    "Percentual (%)": percentual_ausentes
})

tabela_ausentes = tabela_ausentes[
    tabela_ausentes["Valores Ausentes"] > 0
]

tabela_ausentes
```

```
duplicados = df.duplicated().sum()

print("Quantidade de registros duplicados:", duplicados)
print("Percentual de duplicados:", round((duplicados / len(df)) * 100, 2), "%")
```

```
colunas_numericas = [
    "PASSAGEIROS_PAGOS",
    "ASK",
    "RPK",
    "COMBUSTIVEL_LITROS",
    "DISTANCIA_VOADA_KM",
    "DECOLAGENS",
    "ASSENTOS",
    "HORAS_VOADAS"
]

print("Valores negativos encontrados:\n")

for coluna in colunas_numericas:
    negativos = (df[coluna] < 0).sum()
    print(f"{coluna}: {negativos}")
```

```
estatisticas = df[colunas_numericas].describe().T

estatisticas
```

```
estatisticas["amplitude"] = (
    estatisticas["max"] - estatisticas["min"]
)

estatisticas["coef_variacao_%"] = (
    estatisticas["std"] / estatisticas["mean"] * 100
).round(2)

estatisticas
```

```
registros_por_ano = df["ANO"].value_counts().sort_index()

print("Quantidade de registros por ano:")
print(registros_por_ano)
```

```
print("Primeiro ano:", df["ANO"].min())
print("Último ano:", df["ANO"].max())
print("Anos presentes:", sorted(df["ANO"].unique()))
```

```
plt.figure(figsize=(10, 5))

registros_por_ano.plot(kind="bar")

plt.title("Quantidade de Registros por Ano - LATAM/TAM")
plt.xlabel("Ano")
plt.ylabel("Quantidade de registros")
plt.xticks(rotation=0)
plt.tight_layout()

plt.show()
```

```
passageiros_milhoes = passageiros_por_ano / 1_000_000

plt.figure(figsize=(10, 5))

barras = plt.bar(
    passageiros_milhoes.index,
    passageiros_milhoes.values
)

plt.title("Passageiros Pagos por Ano - LATAM/TAM")
plt.xlabel("Ano")
plt.ylabel("Passageiros pagos (milhões)")

for barra, valor in zip(barras, passageiros_milhoes.values):
    plt.text(
        barra.get_x() + barra.get_width()/2,
        barra.get_height(),
        f"{valor:.2f} mi",
        ha="center",
        va="bottom"
    )

plt.tight_layout()
plt.show()
```

```
ocupacao_anual = df.groupby("ANO").agg({
    "ASK": "sum",
    "RPK": "sum"
})

ocupacao_anual["TAXA_OCUPACAO"] = (
    ocupacao_anual["RPK"] /
    ocupacao_anual["ASK"] * 100
)

ocupacao_anual["TAXA_OCUPACAO"] = (
    ocupacao_anual["TAXA_OCUPACAO"].round(2)
)

ocupacao_anual
```

```
plt.figure(figsize=(10, 5))

plt.plot(
    ocupacao_anual.index,
    ocupacao_anual["TAXA_OCUPACAO"],
    marker="o"
)

plt.title("Taxa de Ocupação por Ano - LATAM/TAM")
plt.xlabel("Ano")
plt.ylabel("Taxa de ocupação (%)")
plt.grid(axis="y", alpha=0.3)

for ano, taxa in zip(
    ocupacao_anual.index,
    ocupacao_anual["TAXA_OCUPACAO"]
):
    plt.text(
        ano,
        taxa + 0.4,
        f"{taxa:.2f}%",
        ha="center"
    )

plt.tight_layout()
plt.show()
```

```
eficiencia_anual = df.groupby("ANO").agg({
    "COMBUSTIVEL_LITROS": "sum",
    "PASSAGEIROS_PAGOS": "sum",
    "DISTANCIA_VOADA_KM": "sum"
})

eficiencia_anual["COMBUSTIVEL_POR_PASSAGEIRO"] = (
    eficiencia_anual["COMBUSTIVEL_LITROS"] /
    eficiencia_anual["PASSAGEIROS_PAGOS"]
)

eficiencia_anual["COMBUSTIVEL_POR_KM"] = (
    eficiencia_anual["COMBUSTIVEL_LITROS"] /
    eficiencia_anual["DISTANCIA_VOADA_KM"]
)

eficiencia_anual.round(2)
```

```
plt.figure(figsize=(10, 5))

valores = eficiencia_anual["COMBUSTIVEL_POR_PASSAGEIRO"]

plt.plot(
    valores.index,
    valores.values,
    marker="o"
)

plt.title("Consumo de Combustível por Passageiro - LATAM/TAM")
plt.xlabel("Ano")
plt.ylabel("Litros por passageiro")
plt.grid(axis="y", alpha=0.3)

for ano, valor in zip(valores.index, valores.values):
    plt.text(
        ano,
        valor + 0.5,
        f"{valor:.2f} L",
        ha="center"
    )

plt.tight_layout()
plt.show()
```

```
colunas_outliers = [
    "PASSAGEIROS_PAGOS",
    "ASSENTOS",
    "DECOLAGENS",
    "COMBUSTIVEL_LITROS",
    "DISTANCIA_VOADA_KM"
]

resultado_outliers = []

for coluna in colunas_outliers:
    dados = df[coluna].dropna()

    q1 = dados.quantile(0.25)
    q3 = dados.quantile(0.75)
    iqr = q3 - q1

    limite_inferior = q1 - 1.5 * iqr
    limite_superior = q3 + 1.5 * iqr

    quantidade = (
        (dados < limite_inferior) |
        (dados > limite_superior)
    ).sum()

    resultado_outliers.append([
        coluna,
        q1,
        q3,
        limite_inferior,
        limite_superior,
        quantidade
    ])

tabela_outliers = pd.DataFrame(
    resultado_outliers,
    columns=[
        "Variável",
        "Q1",
        "Q3",
        "Limite Inferior",
        "Limite Superior",
        "Possíveis Outliers"
    ]
)

tabela_outliers.round(2)
```
